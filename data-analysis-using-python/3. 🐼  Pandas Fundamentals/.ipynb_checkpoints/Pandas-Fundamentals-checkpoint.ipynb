# Pandas Fundamentals

Pandas is a powerful Python library used for data analysis and manipulation.  
It provides easy-to-use data structures like **Series** and **DataFrame** for handling structured data efficiently.

With Pandas, we can:
- Load and read datasets
- Clean and preprocess data
- Analyze and summarize information
- Handle missing values
- Perform filtering, grouping, and aggregation

Pandas is widely used in data science, machine learning, and data analytics projects because it simplifies working with tabular data.

# 🔍 2. Exploring Data

_Before diving into analysis or cleaning, it's important to understand the structure and contents of your dataset. Below are some common commands to get an overview:_

- `df.head()`: Display the first 5 rows to get a quick look at the data.
- `df.tail()`: Display the last 5 rows to check data consistency.
- `df.info()`: Summary of the dataframe including data types and non-null counts.
- `df.describe()`: Statistical summary (mean, std, min, max, quartiles) for numeric columns.
- `df.shape`: Returns the number of rows and columns in the dataframe.
- `df.columns`: Lists all column names.
- `df.dtypes`: Shows the data type of each column.
- `df.nunique()`: Number of unique values in each column, useful for understanding categorical variables.
- `df.memory_usage()`: Shows how much memory each column uses.

# 🧹 3. Cleaning Data

_Cleaning data is crucial for accurate analysis. This section covers common cleaning tasks such as handling missing values, duplicates, renaming, type conversion, and applying functions._

---

```python
df.isna().sum()
```

```python
# Fill missing `performance_score` values with the median:

df_filled = df.fillna({'performance_score': df['performance_score'].median()})
```

```python
df.dropna()
```

```python
df_dropped = df.dropna()
```

```python
df_dropped.shape
```

```python
df.duplicated().sum()
```

```python
df_no_dupes = df.drop_duplicates()
```

```python
df_replaced = df.replace({'department': {'HR': 'Human Resources'}})
```

```python
df['salary'] = df['salary'].astype(int)
```

```python
df_renamed = df.rename(columns={'salary': 'annual_salary'})
```

```python
# apply example - add 10 to age
df['age_plus_10'] = df['age'].apply(lambda x: x + 10)
```

```python
# applymap example (only on numeric columns)
df_numeric = df.select_dtypes(include=np.number)
df_numeric_plus_one = df_numeric.applymap(lambda x: x + 1)
```

```python
# map example on department column to get code
dept_map = {'HR': 1, 'Engineering': 2, 'Sales': 3, 'Marketing': 4}
df['dept_code'] = df['department'].map(dept_map)
```

# 🎯 4. Selecting & Filtering Data

This section demonstrates how to select and filter data in a DataFrame using various methods.

---

```python
df.loc[0:5, ['id', 'age', 'department']]      # label based indexing
```

```python
df.iloc[0:5, 0:3]                             # position based indexing
```

```python
df.query('age > 30 and department == "Sales"')  # query rows
```

```python
df.filter(items=['id', 'salary'])             # filter columns by name
```

```python
df.where(df['age'] > 50).head()               # keep values where condition True else NaN
```

```python
df.mask(df['age'] <= 50).head()               # keep values where condition False else NaN
```

```python
df.sample(5)                                  # random sample of rows
```

```python
df.nlargest(3, 'salary')                       # top 3 highest salaries
```

```python
df.nsmallest(3, 'age')                         # 3 youngest employees
```

```python
# Group by department
df.groupby('department')['salary'].mean()
```

```python
# Pivot table example (mean salary per department per performance score)
pd.pivot_table(df, values='salary', index='department', columns='performance_score', aggfunc='mean')
```

```python
# Aggregation with agg()
df.agg({'age': ['min', 'max', 'mean'], 'salary': ['sum', 'mean']})
```

```python
# Cumulative sum and product of salary
df['cumsum_salary'] = df['salary'].cumsum()
df['cumprod_age'] = df['age'].cumprod()
```

# 🛠️ 6. Modifying Data

```python
# Insert a new column
df.insert(1, 'new_col', 'test')
```

```python
# Concatenate DataFrames
df2 = df.head(10).copy()
df_concat = pd.concat([df, df2], ignore_index=True)
```

```python
# Assign to add a calculated column
df = df.assign(salary_k = df['salary'] / 1000)
```

```python
# Melt - long format
df_melt = pd.melt(df, id_vars=['id'], value_vars=['age', 'salary_k'])
```

```python
# Using pivot_table to handle duplicate entries
df_pivot = df_melt.pivot_table(
    index='id',
    columns='variable',
    values='value',
    aggfunc='first'   # or 'sum', 'mean', etc.
)

print(df_pivot)
```

```python
# Explode - create one row per list element
df_explode = pd.DataFrame({'id': [1, 2], 'colors': [['red', 'blue'], ['green']]})
df_exploded = df_explode.explode('colors')
```

# 🔗 7. Merging & Joining

```python
df_left = df.head(10)
df_right = df[['id', 'department']].tail(10)
```

```python
# Merge on 'id'
df_merged = df_left.merge(df_right, on='id', how='inner')
```

```python
# Join (set index first)
df_left = df_left.set_index('id')
df_right = df_right.set_index('id')
df_joined = df_left.join(df_right, lsuffix='_left', rsuffix='_right')
```

```python
# Concatenate (again)
df_concat = pd.concat([df_left, df_right], axis=0)
```
