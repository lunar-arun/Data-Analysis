# Data Generation and Initial Dataset Creation

In this section, we create a synthetic dataset using `pandas` and `numpy`. This dataset simulates employee records with various attributes such as age, salary, department, joining date, and performance score.

---

### Steps:

- **Set a random seed** for reproducibility.
- Create **500 datapoints** with the following fields:
  - `id`: Unique identifier from 1 to 500.
  - `age`: Random integers between 18 and 70.
  - `salary`: Normally distributed values with mean 50,000 and std deviation 15,000.
  - `department`: Randomly assigned from 4 departments.
  - `join_date`: Random dates within 3 years starting from January 1, 2020.
  - `performance_score`: Randomly chosen from 1 to 5, with some missing values (`NaN`).

- Add a few **duplicate rows** intentionally to simulate real-world data cleaning challenges.

---

_**Step 1: Generate the sample data**_

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Create 500 datapoints
n = 500
df = pd.DataFrame({
    'id': range(1, n+1),
    'age': np.random.randint(18, 70, size=n),
    'salary': np.random.normal(50000, 15000, size=n).round(2),
    'department': np.random.choice(['HR', 'Engineering', 'Sales', 'Marketing'], size=n),
    'join_date': pd.to_datetime('2020-01-01') + pd.to_timedelta(np.random.randint(0, 365*3, size=n), unit='D'),
    'performance_score': np.random.choice([np.nan, 1, 2, 3, 4, 5], size=n, p=[0.1, 0.15, 0.2, 0.3, 0.15, 0.1])
})

# Add some duplicates for cleaning demo
df = pd.concat([df, df.iloc[:5]], ignore_index=True)

df.head()
```

# 🧹 3. Cleaning Data

_Cleaning data is crucial for accurate analysis. This section covers common cleaning tasks such as handling missing values, duplicates, renaming, type conversion, and applying functions._

---

```python
df.isna().sum()
```

```python
# Fill missing `performance_score` values with the median:

df_filled = df.fillna({'performance_score': df['performance_score'].median()})
```

```python
df.dropna()
```

```python
df_dropped = df.dropna()
```

```python
df_dropped.shape
```

```python
df.duplicated().sum()
```

```python
df_no_dupes = df.drop_duplicates()
```

```python
df_replaced = df.replace({'department': {'HR': 'Human Resources'}})
```

```python
df['salary'] = df['salary'].astype(int)
```

```python
df_renamed = df.rename(columns={'salary': 'annual_salary'})
```

```python
# apply example - add 10 to age
df['age_plus_10'] = df['age'].apply(lambda x: x + 10)
```

```python
# applymap example (only on numeric columns)
df_numeric = df.select_dtypes(include=np.number)
df_numeric_plus_one = df_numeric.applymap(lambda x: x + 1)
```

```python
# map example on department column to get code
dept_map = {'HR': 1, 'Engineering': 2, 'Sales': 3, 'Marketing': 4}
df['dept_code'] = df['department'].map(dept_map)
```
