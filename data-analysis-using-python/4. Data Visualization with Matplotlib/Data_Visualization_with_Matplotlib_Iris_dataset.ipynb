```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Display plots inside notebook
%matplotlib inline
```

```python
# Load Iris dataset
iris = load_iris()

# Convert to DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add species column
df['species'] = iris.target

# Display first five rows
df.head()
```

```python
# Dataset information
print("Shape of Dataset:", df.shape)

# Statistical summary
df.describe()
```

# Line Plot

```python
# Create a line plot for sepal length

plt.figure(figsize=(8, 5))

plt.plot(df['sepal length (cm)'],
         color='blue',
         linewidth=2)

plt.title('Line Plot of Sepal Length')
plt.xlabel('Index')
plt.ylabel('Sepal Length (cm)')

plt.grid(True)
plt.show()
```

# Scatter Plot

```python
# Scatter plot between sepal length and petal length

plt.figure(figsize=(8, 5))

plt.scatter(
    df['sepal length (cm)'],
    df['petal length (cm)'],
    color='green'
)

plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')

plt.show()
```

# Bar Chart

```python
# Count samples in each species

species_count = df['species'].value_counts()

plt.figure(figsize=(6, 4))

plt.bar(
    species_count.index,
    species_count.values,
    color=['red', 'blue', 'green']
)

plt.title('Species Count')
plt.xlabel('Species')
plt.ylabel('Count')

plt.show()
```

# Histogram

```python
# Histogram of petal length

plt.figure(figsize=(8, 5))

plt.hist(
    df['petal length (cm)'],
    bins=10,
    color='orange',
    edgecolor='black'
)

plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')

plt.show()
```

# Box Plot

```python
# Box plot for all numerical columns

plt.figure(figsize=(10, 6))

plt.boxplot(
    df.iloc[:, :4],
    labels=df.columns[:4]
)

plt.title('Box Plot of Iris Features')

plt.xticks(rotation=15)
plt.show()
```

# Pie Chart

```python
# Pie chart of species distribution

plt.figure(figsize=(6, 6))

plt.pie(
    species_count.values,
    labels=species_count.index,
    autopct='%1.1f%%'
)

plt.title('Species Distribution')

plt.show()
```

# Multiple Subplots

```python
# Create multiple visualizations together

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Histogram
axes[0, 0].hist(df['sepal length (cm)'], color='skyblue')
axes[0, 0].set_title('Histogram')

# Scatter Plot
axes[0, 1].scatter(
    df['sepal length (cm)'],
    df['petal length (cm)'],
    color='green'
)
axes[0, 1].set_title('Scatter Plot')

# Line Plot
axes[1, 0].plot(df['sepal width (cm)'], color='red')
axes[1, 0].set_title('Line Plot')

# Bar Chart
axes[1, 1].bar(
    species_count.index,
    species_count.values,
    color='purple'
)
axes[1, 1].set_title('Bar Chart')

plt.tight_layout()
plt.show()
```
