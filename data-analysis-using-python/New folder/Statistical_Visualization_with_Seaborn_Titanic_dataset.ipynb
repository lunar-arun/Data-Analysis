```python
# Import required libraries

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Display plots inside notebook
%matplotlib inline

# Set seaborn style
sns.set_style("whitegrid")
```

```python
# Load Titanic dataset from seaborn

df = sns.load_dataset('titanic')

# Display first 5 rows
df.head()
```

```python
# Check dataset dimensions

print("Shape of Dataset:", df.shape)

# Display column information
df.info()

# Statistical summary
df.describe()
```

# Count Plot

```python
# Count plot for passenger class

plt.figure(figsize=(8, 5))

sns.countplot(
    x='class',
    data=df,
    palette='Set2'
)

plt.title('Passenger Count by Class')
plt.xlabel('Class')
plt.ylabel('Count')

plt.show()
```

# Survival Distribution

```python
# Count plot showing survival status

plt.figure(figsize=(8, 5))

sns.countplot(
    x='survived',
    data=df,
    palette='coolwarm'
)

plt.title('Survival Distribution')
plt.xlabel('Survived')
plt.ylabel('Count')

plt.show()
```

# Histogram with KDE

```python
# Distribution of passenger age

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='age',
    kde=True,
    color='skyblue'
)

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```

# Box Plot

```python
# Box plot of age across passenger classes

plt.figure(figsize=(8, 5))

sns.boxplot(
    x='class',
    y='age',
    data=df,
    palette='pastel'
)

plt.title('Age Distribution by Class')
plt.xlabel('Class')
plt.ylabel('Age')

plt.show()
```

# Violin Plot

```python
# Violin plot for age distribution by survival

plt.figure(figsize=(8, 5))

sns.violinplot(
    x='survived',
    y='age',
    data=df,
    palette='muted'
)

plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')

plt.show()
```

# Scatter Plot

```python
# Scatter plot between age and fare

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='age',
    y='fare',
    hue='survived',
    data=df
)

plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')

plt.show()
```

# Pair Plot

```python
# Pairwise relationships among numerical features

sns.pairplot(
    df[['age', 'fare', 'parch', 'sibsp', 'survived']],
    hue='survived'
)

plt.show()
```

# Correlation Heatmap

```python
# Select numerical columns

numeric_df = df.select_dtypes(include=['number'])

# Compute correlation matrix
corr_matrix = numeric_df.corr()

# Plot heatmap

plt.figure(figsize=(10, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

# Bar Plot

```python
# Average fare by passenger class

plt.figure(figsize=(8, 5))

sns.barplot(
    x='class',
    y='fare',
    data=df,
    palette='viridis'
)

plt.title('Average Fare by Class')
plt.xlabel('Class')
plt.ylabel('Average Fare')

plt.show()
```

# KDE Plot

```python
# Kernel Density Estimation for age

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x='age',
    fill=True,
    color='purple'
)

plt.title('KDE Plot of Age')
plt.xlabel('Age')
plt.ylabel('Density')

plt.show()
```

# Joint Plot

```python
# Joint plot of age and fare

sns.jointplot(
    data=df,
    x='age',
    y='fare',
    kind='scatter',
    height=7
)

plt.show()
```

# Key observations:

#### 1. Most passengers traveled in Third Class.
#### 2. Younger passengers had slightly higher survival rates.
#### 3. Fare and passenger class are strongly related.
#### 4. Boxplots and violin plots reveal age distribution differences.
#### 5. Heatmaps help identify correlations among numerical variables.
#### 6. Pairplots provide a quick overview of feature relationships.
